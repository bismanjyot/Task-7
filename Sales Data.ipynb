```python
import sqlite3
conn = sqlite3.connect("Sales_Data.db")
query = "SELECT Mode, Count(Customer) FROM Sales GROUP BY Mode"
```

```python
import pandas as pd
df = pd.read_sql_query(query, conn)
print(df)
```

```python
import matplotlib.pyplot as plt 
df.plt(kind='bar', x='Items', y='Count')
plt.savefig("sales_chart.png")
```
